```python
import numpy as np
import matplotlib.pyplot as plt

# Signal frequency
fm = 5

# Continuous-time signal
t = np.linspace(0, 1, 4000)
x = np.sin(2 * np.pi * fm * t)

# Different sampling frequencies
sampling_rates = [20, 10, 7]

# Plotting the signals
fig, axes = plt.subplots(3, 1, figsize=(10, 9))

for ax, fs in zip(axes, sampling_rates):

    # Sampling interval
    Ts = 1 / fs

    # Sampled time values
    ts = np.arange(0, 1 + Ts, Ts)

    # Sampled signal
    xs = np.sin(2 * np.pi * fm * ts)

    # Original signal
    ax.plot(t, x, label="Original Signal")

    # Sampled signal
    ax.stem(ts, xs, linefmt='C1-',
            markerfmt='C1o', basefmt=' ')

    ax.set_title(f"Sampling Frequency = {fs} Hz")
    ax.set_ylabel("Amplitude")
    ax.grid(True)
    ax.legend()

axes[-1].set_xlabel("Time (seconds)")

plt.tight_layout()
plt.show()

# Displaying results
print("Signal Frequency:", fm, "Hz")
print("Nyquist Rate:", 2 * fm, "Hz")

for fs in sampling_rates:
    if fs >= 2 * fm:
        print(fs, "Hz: Nyquist condition satisfied")
    else:
        print(fs, "Hz: Aliasing occurs")
```
